```python
# ============================================================
# 0. PARAMÈTRES : taux de conversion (À AJUSTER si besoin)
# ============================================================
USD_EUR = 0.92        # 1 $  -> €
GBP_EUR = 1.15        # 1 £  -> €
INR_EUR = 0.0105      # 1 ₹  -> €
MILES_KM = 1.609      # 1 mile -> km

COLONNES_FINALES = ["Model", "year", "kilométrage", "fuel_type", "transmission", "prix_euro"]
```

```python
# ============================================================
# 1. FONCTIONS UTILITAIRES (communes aux 4 bases)
# ============================================================
def extraire_nombre(serie):
    """Garde seulement chiffres et point : '51,000 mi.' -> 51000.0 ; '$10,300' -> 10300.0"""
    nettoye = serie.astype(str).str.replace(r"[^\d.]", "", regex=True)
    return pd.to_numeric(nettoye, errors="coerce")

def unifier_carburant(x):
    """Ramène tous les carburants à : petrol, diesel, hybrid, electric (sinon NaN)."""
    if pd.isna(x):
        return np.nan
    x = str(x).lower()
    if "hybrid" in x:                                  # Hybrid, Plug-In Hybrid...
        return "hybrid"
    if "electric" in x:
        return "electric"
    if "diesel" in x:
        return "diesel"
    if "gasoline" in x or "petrol" in x or "flex" in x:   # essence
        return "petrol"
    return np.nan                                      # '–', 'not supported', Other...

def unifier_transmission(x):
    """Ramène toutes les transmissions à : manual, automatic, semi-auto (sinon NaN)."""
    if pd.isna(x):
        return np.nan
    x = str(x).lower()
    if "semi" in x or "dual" in x:                     # Semi-Auto, Semi-Automatic, Dual Shift
        return "semi-auto"
    if "manual" in x or "m/t" in x:                    # Manual, 6-Speed M/T
        return "manual"
    if "auto" in x or "a/t" in x or "cvt" in x:        # Automatic, 8-Speed A/T, CVT
        return "automatic"
    return np.nan

def finaliser(bis):
    """Garde les 6 colonnes finales, dans l'ordre, avec les bons types."""
    bis = bis[COLONNES_FINALES].copy()
    bis["Model"] = bis["Model"].astype(str).str.strip().str.lower()
    bis["year"] = pd.to_numeric(bis["year"], errors="coerce")
    bis["kilométrage"] = pd.to_numeric(bis["kilométrage"], errors="coerce")
    bis["prix_euro"] = pd.to_numeric(bis["prix_euro"], errors="coerce")
    return bis
```

```python
 #============================================================
# 2. DB1 -> db1_bis
# ============================================================
db1_bis = pd.DataFrame()
db1_bis["Model"] = db1["brand"] + " " + db1["model"]              # marque + modèle
db1_bis["year"] = db1["model_year"]
db1_bis["kilométrage"] = extraire_nombre(db1["milage"]) * MILES_KM  # miles -> km
db1_bis["fuel_type"] = db1["fuel_type"].apply(unifier_carburant)
db1_bis["transmission"] = db1["transmission"].apply(unifier_transmission)
db1_bis["prix_euro"] = extraire_nombre(db1["price"]) * USD_EUR      # $ -> €
db1_bis = finaliser(db1_bis)
```

```python
# ============================================================
# 3. DB2 -> db2_bis
# ============================================================
db2_bis = pd.DataFrame()
db2_bis["Model"] = db2["Brand"] + " " + db2["Model"]
db2_bis["year"] = db2["Year"]
db2_bis["kilométrage"] = db2["Mileage"]                 # on suppose déjà en km
db2_bis["fuel_type"] = db2["Fuel_Type"].apply(unifier_carburant)
db2_bis["transmission"] = db2["Transmission"].apply(unifier_transmission)
db2_bis["prix_euro"] = db2["Price"]                     # on suppose déjà en €
db2_bis = finaliser(db2_bis)
```

```python
# ============================================================
# 4. DB3 -> db3_bis  (base 100% Honda)
# ============================================================
db3_bis = pd.DataFrame()
db3_bis["Model"] = db3["Car Model"]                     # ex: "Honda City V CVT"
# Si tu veux seulement "Honda" pour toutes les lignes :
# db3_bis["Model"] = "Honda"
db3_bis["year"] = db3["Year"]
db3_bis["kilométrage"] = extraire_nombre(db3["kms Driven"])          # déjà en km
db3_bis["fuel_type"] = db3["Fuel Type"].apply(unifier_carburant)
db3_bis["transmission"] = db3["Suspension"].apply(unifier_transmission)  # 'Suspension' = transmission
db3_bis["prix_euro"] = extraire_nombre(db3["Price"]) * 100000 * INR_EUR  # Lakh -> ₹ -> €
db3_bis = finaliser(db3_bis)
```

```python
# ============================================================
# 5. DB4 -> db4_bis
# ============================================================
db4_bis = pd.DataFrame()
db4_bis["Model"] = db4["Manufacturer"] + " " + db4["model"]
db4_bis["year"] = db4["year"]
db4_bis["kilométrage"] = db4["mileage"] * MILES_KM      # données UK : miles -> km
db4_bis["fuel_type"] = db4["fuelType"].apply(unifier_carburant)
db4_bis["transmission"] = db4["transmission"].apply(unifier_transmission)
db4_bis["prix_euro"] = db4["price"] * GBP_EUR           # £ -> €
db4_bis = finaliser(db4_bis)
```

```python
# ============================================================
# 6. MERGE -> bd_merged
# ============================================================
bd_merged = pd.concat([db1_bis, db2_bis, db3_bis, db4_bis], ignore_index=True)
print("Après merge :", bd_merged.shape)
```

```python
bd_merged.head()
len(bd_merged)
```

```python
# ============================================================
# 7. NETTOYAGE (simple)
# ============================================================
# 7.1 Doublons
avant = len(bd_merged)
bd_merged = bd_merged.drop_duplicates()
print("Doublons supprimés :", avant - len(bd_merged))

# 7.2 Lignes trop incomplètes (on garde si au moins 5 colonnes sur 6 sont remplies)
avant = len(bd_merged)
bd_merged = bd_merged.dropna(thresh=5)
print("Lignes trop incomplètes supprimées :", avant - len(bd_merged))

# 7.3 Sans prix, la ligne est inutile (c'est la cible à prédire)
avant = len(bd_merged)
bd_merged = bd_merged.dropna(subset=["prix_euro"])
print("Lignes sans prix supprimées :", avant - len(bd_merged))

bd_merged = bd_merged.reset_index(drop=True)
print("Base finale :", bd_merged.shape)
print(bd_merged.isna().sum())     # valeurs manquantes restantes par colonne
bd_merged.head()
```
